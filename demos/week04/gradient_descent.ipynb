```python
import numpy as np
import matplotlib.pyplot as plt

# Dit is onze simpele kostfunctie
def cost_function(x):
    return x**4 - 3*x**3 + 2

# De gradient is hier hardcoded
def gradient(x):
    return 4*x**3 - 9*x**2

# Gradient Descent parameters
learning_rate = 0.001
num_iterations = 1000 # dit is een eerste simpele implementatie
initial_x = -3  # Initiële gok

# Lijstjes om alles bij te houden. Gaan we nodig hebben voor plots.
x_history = []
y_history = []

# Gradient descent, versie met FOR loop en vast aantal iteraties
x = initial_x
for i in range(num_iterations):
    x_history.append(x)
    y = cost_function(x)
    y_history.append(y)
    grad = gradient(x)
    x = x - learning_rate * grad
```

```python
# slimmere versie gradient descent
convergence_threshold = np.sqrt(np.finfo(float).eps) # dit is al iets slimmer
steps = 0

x_history = []
y_history = []

x = initial_x
converged = False
while not converged:
    x_history.append(x)
    y = cost_function(x)
    y_history.append(y)
    grad = gradient(x)
    new_x = x - learning_rate * grad
    steps += 1
    if np.abs(new_x - x) < convergence_threshold:
        converged = True
    x = new_x
```

Hieronder wat code om de situatie visueel voor te stellen .

```python

# Een kwadratische functie
x_range = np.linspace(-1, 4, 400)
y_range = cost_function(x_range)
plt.plot(x_range, y_range, label='Cost Function')

# Plot the gradient descent tussenliggende stappen
plt.scatter(x_history, y_history, c='red', marker='x', label='Gradient Descent Steps')

'''
# Tekst bij laatste punt
plt.annotate(f'Minimum ({x_history[-1]:.2f}, {y_history[-1]:.2f})', 
             xy=(x_history[-1], y_history[-1]), 
             xytext=(x_history[-1] - 3, y_history[-1] + 20),
             arrowprops=dict(arrowstyle='->'))
'''

# Plot labels, legenda
plt.xlabel('Parameter (x)')
plt.ylabel('Functiewaarde')
plt.title('Gradient Descent Optimisation')
plt.legend()

plt.annotate(f'Aantal stappen tot convergentie: {steps}', 
             xy=(x_history[0], y_history[0]), 
             xytext=(x_history[0] - 3, max(y_history) + 100),
             arrowprops=dict(arrowstyle='->'))

plt.grid()
plt.show()
```
